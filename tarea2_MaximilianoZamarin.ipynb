```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random

L_a_W = 3.828e26 # Luminosidad solar en vatios
pc_a_m = 3.085677581491367e16 #parsec en metros


L = [1.0, 2.5, 0.8, 5.0, 10.0] #En unidades de luminosidad Solar
d = [10, 25, 15, 40, 60] #En parsec

#A
luminosidad= np.array(L)
distancia= np.array(d)  

#B
def flujo_estrella(luminosidad,distancia):
    L = np.asarray(luminosidad) * L_a_W      # convertir a vatios
    d = np.asarray(distancia) * pc_a_m       # convertir a metros
    
    return L/(4*np.pi*d**2)

flujos= flujo_estrella(luminosidad, distancia)
print(flujos)

#C-D
indice_maximo= np.argmax(flujos)
indice_minimo= np.argmin(flujos)
print("Indice mayor: ", indice_maximo)
print("Indice menor: ", indice_minimo)

#E
promedio= np.mean(flujos)
mediana= np.median(flujos)
print("Promedio: ", promedio)
print("Mediana: ", mediana)

#F
limite= 10**-11 #En Wm**-2
mascara= flujos>limite #condicional
print("Mascara: ", mascara)

#G
indices_mascara= np.where(mascara)[0]
print("Indices detectados (mascara): ", indices_mascara)

#H
orden=np.argsort(flujos,axis=-1)
print("Orden de mayor a menor flujo: ", orden)

#Nose que comentarios añadir, la mayoria de las funciones np. estan en el material que entrego de la clase 2

    
    
```

```python
import numpy as np
import matplotlib.pyplot as plt

t= np.linspace(0, 10,500) 
flujo= np.ones_like(t)
rng = np.random.default_rng(mi_numero) 
flujo[(t >= 4.5) & (t <= 5.0)] -= 0.02
flujo += rng.normal(0, 0.002, size=t.shape)

#A
flujo_p = np.mean(flujo)
print("Flujo promedio: ", flujo_p)

#B
d_e= np.std(flujo)
print("Desviacion estandar: ", d_e)

#C
flujo_min= np.min(flujo)
indice_min= np.argmin(flujo)
print(f"Flujo minimo: {flujo_min}, Indice minimo: {indice_min}")

#D
tiempo_min= t[indice_min]

#E
umbral= flujo_p - 3*d_e
puntos_transitos = flujo < umbral
print(f"Tiempo del flujo minimo: {tiempo_min} años")

#F
n_puntos = np.sum(puntos_transitos)
print("Puntos de transito detectados: ", n_puntos)

#G
F_tran = np.mean(flujo[puntos_transitos])
delta_F = 1 - F_tran
print("Profundidad del tránsito delta F: ", delta_F)

#H
razon_radios = np.sqrt(delta_F)
print(" Rp / R* =", razon_radios)

#I
tiempos_detectados= t[puntos_transitos]
print("Primer tiempo detectado: ", tiempos_detectados[0])
print("Último tiempo detectado: ", tiempos_detectados[-1])
print("Intervalo real del tránsito: 4.5 – 5.0 años")

```
